```python
#employment trend visualisations
```

```python
for job in df_filtered["Occupation"].unique():
    
    subset = df_filtered[df_filtered["Occupation"] == job]
    
    plt.plot(
        subset["Year"],
        subset["TotalEmployment"],
        marker='o',
        label=job
    )

plt.title("Employment Trends in UK Digital Professions (2021–2024)")
plt.xlabel("Year")
plt.ylabel("Total Employment")
plt.legend()
plt.show()
```

## Initial Observations

- Programmers and software development professionals consistently display the highest employment levels across all selected occupations, suggesting that demand for software development labour has remained strong despite increasing public concern surrounding AI-driven automation.

- Employment in IT business analysis, architecture, and systems design demonstrates relatively stable growth between 2021 and 2024, which may indicate that organisations continue to require workers capable of integrating, managing, and interpreting increasingly complex digital systems.

- Web design professionals represent the smallest occupational category in the dataset, although employment increases noticeably in 2024. This may suggest that AI has not eliminated demand for web design work, but could instead be changing the types of skills required within the profession.

- Graphic and multimedia design occupations remain comparatively stable across the observed period, contradicting more extreme narratives predicting immediate collapse of creative digital work following the emergence of generative AI tools.

- Data analyst employment fluctuates slightly during the period, particularly with a decline in 2023 followed by recovery in 2024. This may reflect broader uncertainty surrounding the integration of AI-assisted analytical tools into data-oriented professions.

- Overall, the selected digital professions do not demonstrate evidence of widespread occupational disappearance between 2021 and 2024. Instead, the findings may support arguments within the literature suggesting that AI is restructuring tasks and workflows rather than fully replacing digital occupations outright.

```python
selected_jobs = [
    "IT business analysts, architects and systems designers",
    "Programmers and software development professionals",
    "Web design professionals",
    "Graphic and multimedia designers",
    "Data analysts"
]

demand_filtered = table3_uk[
    table3_uk["SOC 4 digit label"].isin(selected_jobs)
]
```

```python
demand_filtered = demand_filtered.copy()
```

```python
years = {
    "2017": [col for col in demand_filtered.columns if "17" in str(col)],
    "2018": [col for col in demand_filtered.columns if "18" in str(col)],
    "2019": [col for col in demand_filtered.columns if "19" in str(col)],
    "2020": [col for col in demand_filtered.columns if "20" in str(col)],
    "2021": [col for col in demand_filtered.columns if "21" in str(col)],
    "2022": [col for col in demand_filtered.columns if "22" in str(col)],
    "2023": [col for col in demand_filtered.columns if "23" in str(col)],
    "2024": [col for col in demand_filtered.columns if "24" in str(col)]
}
```

```python
for year, columns in years.items():
    
    demand_filtered[year] = demand_filtered[columns].mean(axis=1)
```

```python
demand_filtered[
    ["SOC 4 digit label", "2021", "2022", "2023", "2024"]
]
```

```python
demand_yearly = demand_filtered.melt(
    id_vars=["SOC 4 digit label"],
    value_vars=["2021", "2022", "2023", "2024"],
    var_name="Year",
    value_name="AverageDemand"
)
```

```python
demand_yearly.head()
```

#Visualisation of Labour Demand Trend

```python
plt.figure(figsize=(12,6))

for job in demand_yearly["SOC 4 digit label"].unique():
    
    subset = demand_yearly[
        demand_yearly["SOC 4 digit label"] == job
    ]
    
    plt.plot(
        subset["Year"],
        subset["AverageDemand"],
        marker='o',
        label=job
    )

plt.title("Labour Demand Trends in UK Digital Professions (2021–2024)")
plt.xlabel("Year")
plt.ylabel("Average Monthly Job Adverts")

plt.legend()
plt.show()
```

## Labour Demand Trend Observations

- Labour demand for programmers and software development professionals declines substantially between 2022 and 2024, despite employment remaining comparatively high in the employment dataset.

- IT business analysts, architects, and systems designers also show falling labour demand after 2022, suggesting reduced hiring activity within parts of the digital labour market.

- In contrast, web design professionals, graphic and multimedia designers, and data analysts display smaller fluctuations, with labour demand remaining relatively more stable.

- The divergence between employment stability and declining labour demand may suggest that AI-driven restructuring affects recruitment patterns before producing large-scale employment decline.

- One possible interpretation is that AI tools increase worker productivity, reducing the need for aggressive new hiring while existing workers remain employed.

- However, broader economic conditions and the post-pandemic slowdown in technology hiring may also contribute to declining labour demand trends, meaning AI cannot be isolated as the sole explanatory factor.

# Regression Analysis of Labour Demand Trends

This section uses multiple linear regression to examine whether labour demand levels are associated with year and occupation type across selected UK digital professions between 2021 and 2024.

The regression model treats average monthly job adverts as the dependent variable, while year and occupation category are used as independent variables.

```python
regression_df = demand_yearly.copy()
```

```python
regression_df["Year"] = regression_df["Year"].astype(int)
```

```python
regression_df["OccupationCode"] = regression_df[
    "SOC 4 digit label"
].astype("category").cat.codes
```

```python
regression_df.head()
```

```python
import statsmodels.api as sm
```

```python
X = regression_df[["Year", "OccupationCode"]]

y = regression_df["AverageDemand"]
```

```python
X = sm.add_constant(X)
```

```python
model = sm.OLS(y, X).fit()

print(model.summary())
```

The quantitative analysis provides partial evidence of labour-market restructuring within selected UK digital professions. While employment levels remain relatively stable or continue growing in several occupations, labour demand trends decline after 2022, particularly for programmers and IT systems professionals.

However, the regression analysis does not produce statistically significant relationships between year, occupation type, and labour demand. This suggests that although descriptive trends indicate changing recruitment patterns, AI-driven restructuring cannot be isolated as the sole explanatory factor within this dataset.

The findings therefore support a more cautious interpretation: AI may contribute to changes in hiring behaviour and productivity expectations without yet producing clear evidence of widespread employment replacement.

# Comparative Analysis of Employment and Labour Demand Trends

This section compares employment trends with labour demand trends across selected UK digital professions between 2021 and 2024. The aim is to investigate whether AI-driven restructuring appears more strongly in recruitment patterns than in overall employment levels.

```python
employment_compare = df_filtered.copy()

employment_compare = employment_compare.rename(
    columns={"Occupation": "Job"}
)

demand_compare = demand_yearly.copy()

demand_compare = demand_compare.rename(
    columns={"SOC 4 digit label": "Job"}
)
```

```python
comparison_df = pd.merge(
    employment_compare,
    demand_compare,
    on=["Job", "Year"]
)
```

```python
comparison_df.head()
```

```python
employment_compare["Job"] = employment_compare["Job"].str.replace(
    r"^\d{4}\s+", 
    "", 
    regex=True
)
```

```python
employment_compare["Year"] = employment_compare["Year"].astype(str)
demand_compare["Year"] = demand_compare["Year"].astype(str)
```

```python
comparison_df = pd.merge(
    employment_compare,
    demand_compare,
    on=["Job", "Year"]
)
```

```python
comparison_df.shape
comparison_df.head()
```

```python
fig, ax1 = plt.subplots(figsize=(12,6))

for job in comparison_df["Job"].unique():
    subset = comparison_df[comparison_df["Job"] == job]
    
    ax1.plot(
        subset["Year"],
        subset["TotalEmployment"],
        marker='o',
        label=f"{job} Employment"
    )

ax1.set_xlabel("Year")
ax1.set_ylabel("Employment")

ax2 = ax1.twinx()

for job in comparison_df["Job"].unique():
    subset = comparison_df[comparison_df["Job"] == job]
    
    ax2.plot(
        subset["Year"],
        subset["AverageDemand"],
        linestyle='--',
        label=f"{job} Demand"
    )

ax2.set_ylabel("Average Monthly Job Adverts")

plt.title("Employment vs Labour Demand in UK Digital Professions (2021–2024)")

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    bbox_to_anchor=(1.05,1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

## Comparative Analysis Observations

- The comparison graph demonstrates a divergence between employment trends and labour demand trends across several UK digital professions.

- Employment levels generally remain stable or continue increasing between 2021 and 2024, particularly for programmers and IT systems professionals.

- In contrast, labour demand declines substantially after 2022, especially within programming and IT systems occupations.

- This suggests that changes in recruitment behaviour may occur before significant declines in total employment levels.

- One possible interpretation is that AI technologies increase productivity within digital professions, reducing firms’ need for aggressive recruitment while existing workers remain employed.

- The findings therefore support the argument that AI-driven restructuring may currently manifest more strongly through hiring slowdowns and changing labour-market dynamics than through immediate large-scale job replacement.
